# Argo Data Exploration
This notebook explores physical oceanography data collected by **Argo floats**. It includes steps for loading the dataset, checking its structure, and plotting vertical profiles of temperature and salinity.

```python
import os
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature

print("Libraries loaded successfully.")
```

## 1. Load the Data
Replace `'path_to_argo_file.nc'` with the actual path to your downloaded Argo NetCDF file.

```python
# Load a single Argo profile NetCDF file
file_path = 'path_to_argo_file.nc'

if os.path.exists(file_path):
    ds = xr.open_dataset(file_path)
    print(ds)
else:
    print(f"File not found at {file_path}. Please update the path.")
```

## 2. Inspect Metadata & Variables
Check the dimensions, coordinates, and available variables (like `TEMP`, `PSAL`, and `PRES`).

```python
if 'ds' in locals():
    # Display coordinate dimensions
    print("Dimensions:", ds.dims)
    
    # Check data variables
    print("\nData Variables:", list(ds.data_vars.keys()))
```

## 3. Plot Vertical Profiles
Visualize how Sea Water Temperature (`TEMP`) and Practical Salinity (`PSAL`) change with Pressure (`PRES`).

```python
if 'ds' in locals():
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6), sharey=True)

    # Plot Temperature vs Pressure
    ax1.plot(ds['TEMP'].values.flatten(), ds['PRES'].values.flatten(), 'b-o', markersize=4)
    ax1.set_xlabel('Temperature (°C)')
    ax1.set_ylabel('Pressure (dbar)')
    ax1.set_title('Temperature Profile')
    ax1.invert_yaxis()  # Ocean pressure increases downward
    ax1.grid(True)

    # Plot Salinity vs Pressure
    ax2.plot(ds['PSAL'].values.flatten(), ds['PRES'].values.flatten(), 'r-o', markersize=4)
    ax2.set_xlabel('Salinity (psu)')
    ax2.set_title('Salinity Profile')
    ax2.grid(True)

    plt.tight_layout()
    plt.show()
```
